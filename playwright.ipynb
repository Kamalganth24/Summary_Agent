```python

from playwright.async_api import async_playwright
import asyncio

async def extract_section_api(url, section):
    captured_response = []
    capture_mode = {"active": False}

    async def handle_response(response):
        if not capture_mode["active"]:
            return

        api_url = response.url

        if "_next/data" in api_url:
            try:
                data = await response.json()
                captured_response.append({
                    "url": api_url,
                    "data": data
                })
                print(f"\n✅ Captured API for section: {section}")
                print("URL:", api_url)

            except:
                pass

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)
        page = await browser.new_page()

        page.on("response", handle_response)

        await page.goto(url)
        await page.wait_for_timeout(3000)

        capture_mode["active"] = True

        try:
            await page.click(f"text={section}")
        except:
            print(f"❌ Could not find section: {section}")
            await browser.close()
            return None

        await page.wait_for_timeout(5000)

        await browser.close()

    if captured_response:
        return captured_response[0]
    else:
        print("❌ No API Captured")
        return None

```
